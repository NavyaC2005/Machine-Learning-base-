LINEAR DATA - HOUSE PRICE PREDICTION

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
```

```python
df=pd.read_csv("house_prices.csv")
```

```python
df
```

```python
df.corr()
```

CONCLUSION  --->  
1. Area Sq Ft has a very strong positive correlation (0.959858) with House Price.
2. Bedrooms and Bathrooms have weak positive correlations, while House Age and Distance to City have weak negative correlations with House Price.
3. Therefore, Area Sq Ft is the most strongly linearly associated feature with House Price in this dataset.


```python
x=df.drop(columns='House Price')
y=df['House Price']
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
y_train_predicted=model.predict(x_train)
y_train_predicted
```

```python
y_test_predicted=model.predict(x_test)
y_test_predicted
```

```python
# r2 score

r2_train_score=r2_score(y_train,y_train_predicted)
r2_train_score
```

1. Conclusion for Training R² Score

The training R² score is 0.9964, indicating that the model explains approximately 99.64% of the variance in the training data. This suggests that the model fits the training data very well.


```python

r2_test_score=r2_score(y_test,y_test_predicted)
r2_test_score
```

2. Conclusion for Testing R² Score

The testing R² score is 0.9934, indicating that the model explains approximately 99.34% of the variance in the testing data. This suggests that the model performs very well on unseen data.


```python
# mse

mse_train_score=mean_squared_error(y_train,y_train_predicted)
mse_train_score
```


3. Conclusion for Training MSE

The training Mean Squared Error (MSE) is 27,093,604.99, representing the average squared difference between the actual and predicted training values. A lower MSE indicates better prediction accuracy.


```python
mse_test_score=mean_squared_error(y_test,y_test_predicted)
mse_test_score
```

4. Conclusion for Testing MSE

The testing Mean Squared Error (MSE) is 41,808,058.45, which is higher than the training MSE. This indicates that the model makes slightly larger prediction errors on unseen data.


```python
# mae
mae_train_score=mean_absolute_error(y_train,y_train_predicted)
mae_train_score
```

5.Conclusion for Training MAE

The training Mean Absolute Error (MAE) is 4126.61, indicating that the model's predictions differ from the actual training values by approximately 4126.61 units on average.


```python
mae_test_score=mean_absolute_error(y_test,y_test_predicted)
mae_test_score
```

5.Conclusion for Testing MAE
 
 The testing Mean Absolute Error (MAE) is 5210.28, indicating that the model's predictions differ from the actual testing values by approximately 5210.28 units on average.


```python
# rmse
rmse_train_score=root_mean_squared_error(y_train,y_train_predicted)
rmse_train_score
```


6. Conclusion for Training RMSE

The training Root Mean Squared Error (RMSE) is 5205.15, representing the typical prediction error in the training data, with greater emphasis on larger errors.


```python
rmse_test_score=root_mean_squared_error(y_test,y_test_predicted)
rmse_test_score
```


7.Conclusion for Testing RMSE

The testing Root Mean Squared Error (RMSE) is 6465.92, which is higher than the training RMSE. This indicates that the model makes somewhat larger prediction errors on unseen data.
