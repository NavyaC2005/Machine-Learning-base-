NON LINEAR DATA - ELECTRICITY


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
```

```python
df=pd.read_csv("electricity_usage.csv")
```

```python
df
```

```python
df.corr()
```

CONCLUSION  ---->

1. AC Usage Hours/Day has the strongest positive correlation (0.738) with Electricity Usage.
2. Home Area and Household Size show weak positive correlations, while Outside Temperature and Appliance Count show weak negative correlations with Electricity Usage.
3. Therefore, AC Usage Hours/Day has the strongest linear relationship with Electricity Usage among the five independent variables.


```python
x=df.drop(columns="Electricity Usage kWh")
x
y=df["Electricity Usage kWh"]
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
y_train_predicted=model.predict(x_train)
y_train_predicted
```

```python
y_test_predicted=model.predict(x_test)
y_test_predicted
```

```python
# r2 score

r2_train_score=r2_score(y_train,y_train_predicted)
r2_train_score
```

1. Conclusion for Training R² Score

The training R² score is 0.8861, indicating that the model explains approximately 88.61% of the variance in the training data. This shows a good fit.

```python
r2_test_score=r2_score(y_test,y_test_predicted)
r2_test_score
```

2. Conclusion for Testing R² Score

The testing R² score is 0.8580, indicating that the model explains approximately 85.80% of the variance in the testing data. This suggests good predictive performance on unseen data.

```python
# mse

mse_train_score=mean_squared_error(y_train,y_train_predicted)
mse_train_score
```

3. Conclusion for Training MSE

The training Mean Squared Error (MSE) is 705.79, representing the average squared difference between actual and predicted training values.

```python
mse_test_score=mean_squared_error(y_test,y_test_predicted)
mse_test_score
```

4. Conclusion for Testing MSE

The testing Mean Squared Error (MSE) is 670.41, which is lower than the training MSE. This indicates slightly smaller squared prediction errors on the testing data.

```python
# mae
mae_train_score=mean_absolute_error(y_train,y_train_predicted)
mae_train_score
```

1. Conclusion for Training MAE

The training Mean Absolute Error (MAE) is 22.03, indicating that the model's predictions differ from the actual training values by approximately 22.03 units on average.

```python
mae_test_score=mean_absolute_error(y_test,y_test_predicted)
mae_test_score
```

2. Conclusion for Testing MAE

The testing Mean Absolute Error (MAE) is 20.18, indicating that the model's predictions differ from the actual testing values by approximately 20.18 units on average.

```python
# rmse
rmse_train_score=root_mean_squared_error(y_train,y_train_predicted)
rmse_train_score
```

3. Conclusion for Training RMSE

The training Root Mean Squared Error (RMSE) is 26.57, indicating the prediction error on the training data, with greater emphasis on larger errors.

```python
rmse_test_score=root_mean_squared_error(y_test,y_test_predicted)
rmse_test_score
```

4. Conclusion for Testing RMSE

The testing Root Mean Squared Error (RMSE) is 25.89, which is slightly lower than the training RMSE. This indicates slightly smaller prediction errors on the testing data.
