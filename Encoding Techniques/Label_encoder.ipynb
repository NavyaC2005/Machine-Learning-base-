### LabelEncoder

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('loan_data (1).csv')
df.head()
```

```python
df.info()
```

```python
X = df.drop(columns='loan_status')
y = df['loan_status']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(
    X,
    y,
    train_size=0.8,
    random_state=42
)
```

```python
X
```

```python
cat_cols = X.select_dtypes(include='object')
cat_cols.nunique()
```

```python
cat_cols.head()
```

```python
xtrain
```

```python
'''.
label_encoder = LabelEncoder()
encoded_values_train = label_encoder.fit_transform(xtrain) 
Here xtrain contains 13 columns which includes both numerical and 
categorical columns but fit_transform accepts only one categorical column.
# ValueError: y should be a 1d array, got an array of shape (36000, 13) instead.
'''
#correct syntax
label_encoder = LabelEncoder()
encoded_values_train = label_encoder.fit_transform(xtrain['person_gender'])
encoded_values_train 
```

```python
encoded_values_test = label_encoder.transform(xtest['person_gender'])
# if we pass two or more categorical columns => ValueError
encoded_values_test 
```
